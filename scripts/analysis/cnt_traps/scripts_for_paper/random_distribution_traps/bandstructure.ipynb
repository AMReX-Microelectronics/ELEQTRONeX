```python
def Get_MeanAndStd(Var, valid_steps=None):
    """
    Given a variable with shape (num_folders, num_steps, pts), this function
    returns the mean, low, and high values over the folders, as well as their 
    standard deviation.
    
    Args:
    - Var: NumPy array of shape (num_folders, num_steps, pts)
    
    Returns:
    - NewVar: Array of shape (3, num_steps, pts) containing the mean, low, 
      and high values.
    """
    n_folders, n_steps, n_pts = Var.shape
    if valid_steps is None:
        valid_steps = n_steps
        
    #NewVar = np.zeros((3, n_steps, n_pts), dtype=float)
    
    # Compute mean and stddev along the folder axis
    mean = np.mean(Var, axis=0)  # shape (num_steps, pts)
    stddev = np.std(Var, axis=0)  # shape (num_steps, pts)
    
    # Calculate hi and lo using the standard deviation
    hi = mean + 1 * stddev
    lo = mean - 1 * stddev
    
    # Pack them into the result array
    # NewVar[0] = mean  # mean
    # NewVar[1] = lo    # lower bound
    # NewVar[2] = hi    # upper bound

    # assert NewVar.shape == (3, n_steps, n_pts), "Shape mismatch"
    
    return mean[:valid_steps], lo[:valid_steps], hi[:valid_steps]
```
